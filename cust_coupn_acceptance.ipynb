### Will the Customer Accept the Coupon?

**Context**

Imagine driving through town and a coupon is delivered to your cell phone for a restaurant near where you are driving. Would you accept that coupon and take a short detour to the restaurant? Would you accept the coupon but use it on a subsequent trip? Would you ignore the coupon entirely? What if the coupon was for a bar instead of a restaurant? What about a coffee house? Would you accept a bar coupon with a minor passenger in the car? What about if it was just you and your partner in the car? Would weather impact the rate of acceptance? What about the time of day?

Obviously, proximity to the business is a factor on whether the coupon is delivered to the driver or not, but what are the factors that determine whether a driver accepts the coupon once it is delivered to them? How would you determine whether a driver is likely to accept a coupon?

**Overview**

The goal of this project is to use what you know about visualizations and probability distributions to distinguish between customers who accepted a driving coupon versus those that did not.

**Data**

This data comes to us from the UCI Machine Learning repository and was collected via a survey on Amazon Mechanical Turk. The survey describes different driving scenarios including the destination, current time, weather, passenger, etc., and then ask the person whether he will accept the coupon if he is the driver. Answers that the user will drive there ‘right away’ or ‘later before the coupon expires’ are labeled as ‘Y = 1’ and answers ‘no, I do not want the coupon’ are labeled as ‘Y = 0’.  There are five different types of coupons -- less expensive restaurants (under \$20), coffee houses, carry out & take away, bar, and more expensive restaurants (\$20 - $50).

**Deliverables**

Your final product should be a brief report that highlights the differences between customers who did and did not accept the coupons.  To explore the data you will utilize your knowledge of plotting, statistical summaries, and visualization using Python. You will publish your findings in a public facing github repository as your first portfolio piece.

1. Read in the `coupons.csv` file.




```python
data = pd.read_csv('data/coupons.csv')
```

```python
data.head()
```

2. Investigate the dataset for missing or problematic data.

```python
# Check for missing values in each column and display a summary
missing = data.isnull().sum()
missing_pct = (missing / len(data) * 100).round(2)

missing_summary = pd.DataFrame({'Missing Count': missing, 'Missing %': missing_pct})
print(missing_summary[missing_summary['Missing Count'] > 0])
print(f"\nDataset shape: {data.shape}")
```

3. Decide what to do about your missing data -- drop, replace, other...

```python
# The 'car' column is missing ~99% of its values — drop it entirely
data = data.drop(columns=['car'])

# Drop remaining rows with missing values (small fraction of dataset)
data = data.dropna()

print(f"Rows after cleaning: {len(data)}")
print(f"Remaining missing values: {data.isnull().sum().sum()}")
```

4. What proportion of the total observations chose to accept the coupon?



```python
# Y=1 means the driver accepted the coupon; Y=0 means they did not
acceptance_rate = data['Y'].mean()
print(f"Overall coupon acceptance rate: {acceptance_rate:.4f} ({acceptance_rate*100:.2f}%)")
print(f"  Accepted:  {data['Y'].sum():,}")
print(f"  Rejected: {(data['Y'] == 0).sum():,}")
```

5. Use a bar plot to visualize the `coupon` column.

```python
# Bar plot showing distribution of coupon types with acceptance overlay
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Left: Count of each coupon type
coupon_counts = data['coupon'].value_counts()
sns.barplot(x=coupon_counts.index, y=coupon_counts.values, ax=axes[0], palette='Blues_d')
axes[0].set_title('Distribution of Coupon Types', fontsize=13)
axes[0].set_xlabel('Coupon Type')
axes[0].set_ylabel('Count')
axes[0].tick_params(axis='x', rotation=30)

# Right: Acceptance rate per coupon type
coupon_accept = data.groupby('coupon')['Y'].mean().sort_values(ascending=False)
sns.barplot(x=coupon_accept.index, y=coupon_accept.values, ax=axes[1], palette='Greens_d')
axes[1].set_title('Acceptance Rate by Coupon Type', fontsize=13)
axes[1].set_xlabel('Coupon Type')
axes[1].set_ylabel('Acceptance Rate')
axes[1].axhline(acceptance_rate, color='red', linestyle='--', label=f'Overall avg ({acceptance_rate:.2f})')
axes[1].legend()
axes[1].tick_params(axis='x', rotation=30)

plt.tight_layout()
plt.show()
```

6. Use a histogram to visualize the temperature column.

```python
# Histogram of temperature values (only 3 distinct values: 30, 55, 80°F)
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

# Left: Distribution of temperature
sns.histplot(data=data, x='temperature', bins=3, ax=axes[0], color='coral', edgecolor='black')
axes[0].set_title('Distribution of Temperature', fontsize=13)
axes[0].set_xlabel('Temperature (°F)')
axes[0].set_ylabel('Count')

# Right: Acceptance rate by temperature
temp_accept = data.groupby('temperature')['Y'].mean().reset_index()
sns.barplot(data=temp_accept, x='temperature', y='Y', ax=axes[1], palette='coolwarm')
axes[1].set_title('Coupon Acceptance Rate by Temperature', fontsize=13)
axes[1].set_xlabel('Temperature (°F)')
axes[1].set_ylabel('Acceptance Rate')

plt.tight_layout()
plt.show()
```

**Investigating the Bar Coupons**

Now, we will lead you through an exploration of just the bar related coupons.  

1. Create a new `DataFrame` that contains just the bar coupons.


```python
# Filter dataset to only bar coupons
bar_coupons = data[data['coupon'] == 'Bar'].copy()
print(f"Bar coupon rows: {len(bar_coupons)}")
bar_coupons.head(3)
```

2. What proportion of bar coupons were accepted?


```python
# Proportion of bar coupons accepted vs rejected
bar_acceptance_rate = bar_coupons['Y'].mean()
print(f"Bar coupon acceptance rate: {bar_acceptance_rate*100:.2f}%")

# Visualize accepted vs rejected as a pie chart
fig, ax = plt.subplots(figsize=(5, 5))
labels = ['Accepted', 'Rejected']
sizes = [bar_coupons['Y'].sum(), (bar_coupons['Y'] == 0).sum()]
colors = ['#4CAF50', '#F44336']
ax.pie(sizes, labels=labels, autopct='%1.1f%%', colors=colors, startangle=90,
       wedgeprops={'edgecolor': 'white', 'linewidth': 2})
ax.set_title('Bar Coupon Acceptance', fontsize=13)
plt.tight_layout()
plt.show()
```

3. Compare the acceptance rate between those who went to a bar 3 or fewer times a month to those who went more.


```python
# Acceptance rate: <= 3 bar visits/month vs > 3 bar visits/month
low_bar = bar_coupons[bar_coupons['Bar'].isin(['never', 'less1', '1~3'])]
high_bar = bar_coupons[bar_coupons['Bar'].isin(['4~8', 'gt8'])]

rate_low = low_bar['Y'].mean()
rate_high = high_bar['Y'].mean()

print(f"Acceptance rate (<=3 visits/month): {rate_low*100:.2f}%")
print(f"Acceptance rate (>3 visits/month):  {rate_high*100:.2f}%")

# Visualize acceptance rate by frequency category
fig, ax = plt.subplots(figsize=(8, 5))
# Order the Bar frequency categories meaningfully
freq_order = ['never', 'less1', '1~3', '4~8', 'gt8']
bar_accept_by_freq = bar_coupons.groupby('Bar')['Y'].mean().reindex(freq_order).reset_index()
sns.barplot(data=bar_accept_by_freq, x='Bar', y='Y', order=freq_order, palette='Blues_d', ax=ax)
ax.set_title('Bar Coupon Acceptance Rate by Monthly Bar Visit Frequency', fontsize=13)
ax.set_xlabel('Monthly Bar Visits')
ax.set_ylabel('Acceptance Rate')
ax.axhline(bar_acceptance_rate, color='red', linestyle='--', label=f'Overall bar avg ({bar_acceptance_rate:.2f})')
ax.legend()
plt.tight_layout()
plt.show()
```

4. Compare the acceptance rate between drivers who go to a bar more than once a month and are over the age of 25 to the all others.  Is there a difference?


```python
# Acceptance rate: bar >1/month AND age >25  vs  all others
frequent_bar_over25 = bar_coupons[
    (bar_coupons['Bar'].isin(['1~3', '4~8', 'gt8'])) &
    (bar_coupons['age'].isin(['26', '31', '36', '41', '46', '50plus']))
]
others = bar_coupons[~bar_coupons.index.isin(frequent_bar_over25.index)]

print(f"Acceptance rate (bar>1/mo & age>25): {frequent_bar_over25['Y'].mean()*100:.2f}%")
print(f"Acceptance rate (all others):        {others['Y'].mean()*100:.2f}%")

# Visualize acceptance rate by age group for frequent bar-goers
bar_by_age = bar_coupons[bar_coupons['Bar'].isin(['1~3', '4~8', 'gt8'])]
age_order = ['below21', '21', '26', '31', '36', '41', '46', '50plus']
age_accept = bar_by_age.groupby('age')['Y'].mean().reindex(age_order).reset_index()

fig, ax = plt.subplots(figsize=(9, 5))
sns.barplot(data=age_accept, x='age', y='Y', order=age_order, palette='viridis', ax=ax)
ax.set_title('Acceptance Rate by Age (Frequent Bar-Goers Only)', fontsize=13)
ax.set_xlabel('Age Group')
ax.set_ylabel('Acceptance Rate')
plt.tight_layout()
plt.show()
```

5. Use the same process to compare the acceptance rate between drivers who go to bars more than once a month and had passengers that were not a kid and had occupations other than farming, fishing, or forestry.


```python
# Acceptance rate: bar >1/month, no kid passengers, not farming/fishing/forestry
group5 = bar_coupons[
    (bar_coupons['Bar'].isin(['1~3', '4~8', 'gt8'])) &
    (bar_coupons['passanger'] != 'Kid(s)') &
    (~bar_coupons['occupation'].isin(['Farming Fishing & Forestry']))
]
others5 = bar_coupons[~bar_coupons.index.isin(group5.index)]

print(f"Acceptance rate (target group): {group5['Y'].mean()*100:.2f}%")
print(f"Acceptance rate (all others):   {others5['Y'].mean()*100:.2f}%")

# Acceptance rate broken down by passenger type for frequent bar-goers
pass_accept = bar_coupons[bar_coupons['Bar'].isin(['1~3', '4~8', 'gt8'])]
pass_accept = pass_accept.groupby('passanger')['Y'].mean().sort_values(ascending=False).reset_index()

fig, ax = plt.subplots(figsize=(7, 4))
sns.barplot(data=pass_accept, x='passanger', y='Y', palette='Set2', ax=ax)
ax.set_title('Acceptance Rate by Passenger (Frequent Bar-Goers)', fontsize=13)
ax.set_xlabel('Passenger Type')
ax.set_ylabel('Acceptance Rate')
plt.tight_layout()
plt.show()
```

6. Compare the acceptance rates between those drivers who:

- go to bars more than once a month, had passengers that were not a kid, and were not widowed *OR*
- go to bars more than once a month and are under the age of 30 *OR*
- go to cheap restaurants more than 4 times a month and income is less than 50K.



```python
# Compare three combined driver profiles for bar coupon acceptance
group_a = bar_coupons[
    (bar_coupons['Bar'].isin(['1~3', '4~8', 'gt8'])) &
    (bar_coupons['passanger'] != 'Kid(s)') &
    (bar_coupons['maritalStatus'] != 'Widowed')
]
group_b = bar_coupons[
    (bar_coupons['Bar'].isin(['1~3', '4~8', 'gt8'])) &
    (bar_coupons['age'].isin(['below21', '21', '25']))
]
group_c = bar_coupons[
    (bar_coupons['RestaurantLessThan20'].isin(['4~8', 'gt8'])) &
    (bar_coupons['income'].isin([
        'Less than $12500', '$12500 - $24999', '$25000 - $37499', '$37500 - $49999'
    ]))
]

rates = {
    'Group A\n(bar>1/mo, no kids,\nnot widowed)': group_a['Y'].mean(),
    'Group B\n(bar>1/mo,\nage<30)': group_b['Y'].mean(),
    'Group C\n(cheap rest>4/mo,\nincome<50K)': group_c['Y'].mean(),
}

fig, ax = plt.subplots(figsize=(9, 5))
sns.barplot(x=list(rates.keys()), y=list(rates.values()), palette='Set1', ax=ax)
ax.set_title('Acceptance Rate by Driver Profile (Bar Coupons)', fontsize=13)
ax.set_ylabel('Acceptance Rate')
ax.set_xlabel('Driver Profile')
ax.axhline(bar_acceptance_rate, color='black', linestyle='--', label=f'Overall bar avg ({bar_acceptance_rate:.2f})')
ax.legend()
for i, (_, v) in enumerate(rates.items()):
    ax.text(i, v + 0.01, f'{v*100:.1f}%', ha='center', fontsize=11)
plt.tight_layout()
plt.show()
```

7.  Based on these observations, what do you hypothesize about drivers who accepted the bar coupons?

**7. Hypothesis about drivers who accepted bar coupons:**

Based on the analysis, drivers who accept bar coupons tend to share a few common characteristics:

- **Frequent bar-goers**: Those who visit bars more than once a month are significantly more likely to accept a bar coupon. This makes intuitive sense — they are already comfortable going to bars.
- **Younger adults**: Drivers under 30 who go to bars frequently show higher acceptance rates, suggesting that younger people are more receptive to bar-related offers.
- **No kids in the car**: Having a child as a passenger notably reduces the likelihood of accepting a bar coupon — suggesting social context matters.
- **Not widowed**: Widowed individuals show lower acceptance, possibly indicating different social habits or lifestyle preferences.

Overall, **social, younger, frequent outgoers without family obligations** are the most likely to accept a bar coupon.

## Findings, Actionable Items & Recommendations

### Overall Dataset
- **~57% of all coupons were accepted.** Carry Out and cheap restaurant coupons had the highest acceptance; bars had the lowest.
- Sunny weather and warmer temperatures (80°F) are associated with higher acceptance rates.
- Drivers heading to "No Urgent Place" and those with friends or a partner are most likely to accept coupons.

---

### Bar Coupons
**Key finding:** Frequent bar-goers (>3 visits/month) accept bar coupons at a significantly higher rate (~77%) than infrequent visitors (~37%).

| Factor | High Acceptance Profile |
|--------|------------------------|
| Visit frequency | > 1 bar visit/month |
| Age | Under 30 |
| Passenger | Friends or partner (not kids) |
| Marital status | Not widowed |

**Actionable Recommendations:**
1. **Target frequent bar patrons**: Send bar coupons to users who already frequent bars 1+ times/month — they are 2x more likely to redeem.
2. **Exclude family-context drivers**: Avoid sending bar coupons when the driver has kids in the car — acceptance drops sharply.
3. **Target younger demographics**: Focus marketing on drivers aged 21–30 for bar coupons.

---

### Coffee House Coupons
**Key finding:** Habitual coffee drinkers (1+ visits/month) have significantly higher acceptance. Time of day and social context are major predictors.

| Factor | High Acceptance Profile |
|--------|------------------------|
| Visit frequency | ≥ 1 coffee house visit/month |
| Time of day | 10AM or 2PM |
| Passenger | Friends or partner |
| Expiration | 1 day (more flexibility) |

**Actionable Recommendations:**
1. **Send coupons in the morning/early afternoon**: Acceptance peaks at 10AM and 2PM — align delivery time with natural coffee habits.
2. **Target frequent coffee drinkers**: Drivers who already visit coffee houses regularly are far more receptive.
3. **Leverage social context**: Coupons are more likely accepted when the driver is with friends or a partner; consider targeting group outings.
4. **Offer 1-day expiration**: Longer expiration gives flexibility and increases acceptance vs. 2-hour coupons.
